```python
#File Handling in Python
file = open("file1.txt","w")
file.write("I created a File")
print("File is created")
file.close()
```

```python
with open("file1.txt","r") as f:
  data = f.read()
  print(data)
```

```python
with open("file1.txt","a") as f:
  f.writelines("\napple \njuice \nIcecream")
  print("Data is appended")
```

```python
with open("file1.txt","r") as f:
  data=f.readlines()
  print(data)
```

```python
#Coping content of one file to another
with open("file1.txt","r") as f:
  data = f.read()
  print("Reading done")
with open("file2.txt","w") as f:
  f.write(data)
  print("Coping done")
```

```python
#Program to print each line in reverse order
with open("file1.txt","r") as f:
  data = f.readlines()
  for i in data:
    print(i[::-1])
```
