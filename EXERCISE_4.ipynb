```python
#Write a program that counts no.of VOWELS in a String
a = input()
sum = 0
for i in range(len(a)):
  if(a[i]=='a' or a[i] == 'A'or
     a[i]=='e' or a[i] =='E' or
     a[i]=='i' or a[i] =='I' or
     a[i]=='o' or a[i] == 'O'or
     a[i]=='u' or a[i] == 'U'):
    sum += 1
print(sum)
```

```python
def reverse(n):
  r=0
  sum=0
  while(n!=0):
    r = n%10
    sum = sum*10+ r
    n=n//10
  return sum
n = int(input())
x=reverse(n)
print(x)
```
