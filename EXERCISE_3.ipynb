```python
#Program that prints sum of the list elements without using sum()
#Program to print elements in odd index
#Program to print even elements
l1 = [11,55,9,54,63,95,11]
sum=0
for i in l1:
  sum +=i
print(sum,"is the sum of the elements in the list")
print("ODD index elements",end=" : ")
for i in range(len(l1)):
  if(i%2==1):
    print(l1[i],end=" ")
print()
print("Even elements in the list : ",end=" ")
for i in l1:
  if(i%2==0):
    print(i,end=" ")
print()
```
